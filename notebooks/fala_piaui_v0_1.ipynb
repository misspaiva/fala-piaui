## First linguistic inventory

This first version contains five expressions from different linguistic and social dimensions:

- sensory experience;
- affection and interpersonal relationships;
- orientation and disorientation;
- urgency and instruction;
- rural identity and social judgment.

The categories are descriptive labels for this initial collection. They do not represent the frequency of these experiences or expressions across Piauí.

```python
inventory_columns = [
    "expression",
    "word_class",
    "related_forms",
    "category",
    "confidence"
]

linguistic_inventory = df[inventory_columns].sort_values("expression")

linguistic_inventory
```

## Coverage of linguistic dimensions

This chart counts records by descriptive category in the initial collection.

It does not show the frequency of expressions in Piauí or Northeast Brazil. Each bar represents one manually curated entry in version 0.1.

```python
category_counts = df["category"].value_counts().sort_values()

category_counts
```

```python
%pip install matplotlib
```

```python
import matplotlib.pyplot as plt

ax = category_counts.plot(
    kind="barh",
    color="#B5179E",
    figsize=(10, 5)
)

ax.set_title("Fala, Piauí v0.1 — Linguistic dimensions represented")
ax.set_xlabel("Number of records in this initial collection")
ax.set_ylabel("")

plt.tight_layout()
plt.show()
```

## Initial findings and limitations

### What this first collection shows

The five initial records cover distinct linguistic and social dimensions:

- `gastura` documents sensory discomfort and embodied experience;
- `dengo` documents affection, care, and context-dependent interpersonal tone;
- `ariado` documents orientation, disorientation, and familiarity with space;
- `avia` documents urgency, direct instruction, and speaker-listener relationships;
- `matuto` documents rural identity, self-irony, humor, and possible social judgment.

The collection demonstrates that regional language cannot be reduced to isolated “slang words.” Each expression depends on context, relationship, tone, local history, and the speaker's lived experience.

### What this first collection does not show

This notebook does not measure:

- how frequently an expression is used in Piauí;
- whether every person in a city, state, or region recognizes the same meaning;
- whether an expression belongs exclusively to Piauí or Northeast Brazil;
- how meanings differ across generations, cities, social groups, or families;
- a complete linguistic description of Piauí Portuguese.

### Next research questions

Future versions may investigate:

1. Which expressions have no direct English equivalent?
2. Which meanings depend most strongly on tone or speaker-target relationship?
3. Which records need contextual sources or contributions from additional speakers?
4. How do expressions vary across cities and generations within Piauí?
5. How can linguistic variation be documented without stereotyping speakers or communities?
