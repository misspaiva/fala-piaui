## Initial findings and limitations

### What this initial collection shows

The ten current records cover distinct linguistic, social, environmental, and conversational dimensions:

- `gastura` documents sensory discomfort and embodied experience;
- `dengo` documents affection, care, and context-dependent interpersonal tone;
- `ariado` documents orientation, disorientation, and familiarity with space;
- `avia` documents urgency, direct instruction, and speaker-listener relationships;
- `matuto` documents rural identity, self-irony, humor, and possible social judgment;
- `arrodear` documents physical detours and indirect communication;
- `arreado` documents romantic attachment and emotional intensity;
- `valha` documents emotional reaction, surprise, and expressive conversational rhythm;
- `sol esfriar` documents climate as a practical marker of everyday time;
- `oxente` documents how tone, context, and speaker-listener relationships can shape meaning.

The collection demonstrates that regional language cannot be reduced to isolated “slang words.” Expressions may depend on context, relationship, tone, local history, embodied experience, environment, and the speaker's lived linguistic world.

This initial collection is grounded in the author's lived linguistic context in Southern Piauí, in a border area with Pernambuco and Ceará.

### What this initial collection does not show

This notebook does not measure:

- how frequently an expression is used in Southern Piauí, elsewhere in Piauí, or in Northeast Brazil;
- whether every person in a city, state, or region recognizes the same meaning;
- whether an expression documented in a Southern Piauí border context is recognized elsewhere in the state;
- whether an expression belongs exclusively to Piauí, Northeast Brazil, or any single region;
- how vocabulary, pronunciation, meaning, and tone differ between Northern, Central, and Southern Piauí;
- how meanings differ across generations, cities, families, rural and urban contexts, or social groups;
- a complete linguistic description of Portuguese as spoken in Piauí;
- clinical, educational, social, or demographic conclusions about speakers.
